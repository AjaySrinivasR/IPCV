```python
import cv2 as cv

img = cv.imread('messi5.jpg', cv.IMREAD_GRAYSCALE)

assert img is not None, "file could not be read, check with os.path.exists()"

rows, cols = img.shape

# cols-1 and rows-1 are the coordinate limits
M = cv.getRotationMatrix2D(((cols-1)/2.0, (rows-1)/2.0), 90, 1)

dst = cv.warpAffine(img, M, (cols, rows))

from google.colab.patches import cv2_imshow
cv2_imshow(dst)
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img = cv.imread('drawing.pnG')

assert img is not None, "file could not be read, check with os.path.exists()"

rows, cols, ch = img.shape

# Three points in the input image
pts1 = np.float32([[50,50], [200,50], [50,200]])

# Corresponding points in the output image
pts2 = np.float32([[10,100], [200,50], [100,250]])

# Affine transformation matrix
M = cv.getAffineTransform(pts1, pts2)

# Apply affine transform
dst = cv.warpAffine(img, M, (cols, rows))

# Display images
plt.subplot(121), plt.imshow(img), plt.title('Input')
plt.subplot(122), plt.imshow(dst), plt.title('Output')

plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img = cv.imread('sudoku.png')

assert img is not None, "file could not be read, check with os.path.exists()"

rows, cols, ch = img.shape

# Points in original image
pts1 = np.float32([[56,65], [368,52], [28,387], [389,390]])

# Points in output image
pts2 = np.float32([[0,0], [300,0], [0,300], [300,300]])

# Perspective transformation matrix
M = cv.getPerspectiveTransform(pts1, pts2)

# Apply perspective transform
dst = cv.warpPerspective(img, M, (300,300))

# Display images
plt.subplot(121), plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB)), plt.title('Input')
plt.subplot(122), plt.imshow(cv.cvtColor(dst, cv.COLOR_BGR2RGB)), plt.title('Output')

plt.show()
```
